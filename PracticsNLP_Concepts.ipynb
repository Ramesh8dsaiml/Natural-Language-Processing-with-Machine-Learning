```python
import re
import string

def clean_text(text):
    
    # 1. Lowercase
    text = text.lower()

    # 2. Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # 3. Remove HTML Tags
    text = re.sub(r'<.*?>', '', text)

    # 4. Remove Numbers
    text = re.sub(r'\d+', '', text)

    # 5. Remove Punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # 6. Remove Extra Spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

text = """
<h1>Python 2026</h1>

Visit https://openai.com

Python is AWESOME!!!!
"""
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

lemmatizer = WordNetLemmatizer()

text = "The children were running faster than the boys and ate apples."

tokens = word_tokenize(text)

result = []

for word in tokens:

    if word.lower() in ["Running", "Were", "ate"]:
        result.append(lemmatizer.lemmatize(word, pos="v"))

    elif word.lower() == "faster":
        result.append(lemmatizer.lemmatize(word, pos="a"))

    else:
        result.append(lemmatizer.lemmatize(word))

print("Original Tokens")
print(tokens)

print("\nLemmatized Tokens")
print(result)
```

# One Hot Encoding?
one hot encoding  is a Technique used to convert categorial values ( or  words ) in to a binary   vector ( 0 ,1 )

## why called one hot ?
every  vector have only one  value(1) otherwise zero (0)


```python
# manual :
sentence = "I love Python"

words = sentence.split()

vocabulary = sorted(set(words))

print("Vocabulary:", vocabulary)

for word in vocabulary:

    vector = []

    for vocab_word in vocabulary:

        if word == vocab_word:
            vector.append(1)
        else:
            vector.append(0)

    print(word, "->", vector)
    
```

# N-Grams:
N-Gram is a sequence of N consecutive words from a  given tex.

Simple Language:

Agar hum 1 word ek saath lete hain → Unigram,

Agar 2 words ek saath lete hain → Bigram,

Agar 3 words ek saath lete hain → Trigram,

## Example Sentence

I love Machine Learning

Unigram (1 Word)

I

love

Learning

## Bigram (2 Words)

I love

Machine Learning

## Trigram (3 Words)

I love Machine

love Machine Learning




CountVectorizer me N-Gram

CountVector izer(ngram_range=(1,1))

Bigram

CountVectorizer(ngram_range=(2,2))


Trigram

CountVectorizer(ngram_range=(3,3))

Unigram + Bigram

CountVectorizer(ngram_range=(1,2))

Unigram + Bigram + Trigram

CountVectorizer(ngram_range=(1,3))

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

docs = [
    "I love Python",
    "Python is easy",
    "what are you doing ",
            "tell me about your self ",
            "are you ready for rashling",
            "will you marry me "
]

cv = CountVectorizer(ngram_range=(1,1))

X = cv.fit_transform(docs)

df = pd.DataFrame(
    X.toarray(),
    columns=cv.get_feature_names_out()
)

print(df)
```

# TF = Term Frequency

Simple meaning

Ek document me koi word kitni baar aaya.

# Formula

TF = (Word Count) / (Total Words in Document)

# What is IDF?

Inverse Document Frequency

IDF = log(Total Documents / Documents Containing Word)

# TF-IDF Formula

TF-IDF = TF × IDF

```python
# Word

# ↓

# TF

# (Document me kitni baar)

# ↓

# IDF

# (Pure Corpus me kitna rare)

# ↓

# TF × IDF

# ↓

# Importance Score
```
